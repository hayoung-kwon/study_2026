```python
from PIL import Image
import numpy as np
import pandas as  pd

# main idea : 실제 픽셀 데이터를 numpy 배열로 변환하여 계산한다

img1 = Image.open("apple.jpg")
img2 = Image.open("apple2.jpg")
img3 = Image.open("apple3.jpg")

# scaling (이미지 크기와 색상 형식 통일)
image_size=(128, 128)
img1_resized = img1.convert("RGB").resize(image_size)
img2_resized = img2.convert("RGB").resize(image_size)
img3_resized = img3.convert("RGB").resize(image_size)

# numpy 배열로 변환하고 0~1 범위로 scaling
array1 = np.array(img1_resized).astype(float)/255.0
array2 = np.array(img2_resized).astype(float)/255.0
array3 = np.array(img3_resized).astype(float)/255.0

# 이미지를 1차원 벡터로 변환
# 이유 : 이미지 하나를 하나의 벡터로 표현해서 이미지 간의 유사도를 비교하기위해
# 원래 128*128*3인데 그걸 일렬로 눌러서 하나의 벡터로 만든다
vector1 = array1.flatten()
vector2 = array2.flatten()
vector3 = array3.flatten()

def cosine_similarity(vector_a, vector_b) :
  dot_product = np.dot(vector_a, vector_b)
  magnitude = (np.linalg.norm(vector_a)*np.linalg.norm(vector_b))
  if magnitude == 0 :
    return 0
  return dot_product/magnitude

# 정의한 함수로 코사인 유사도 다 계산해놓기
sim12 = cosine_similarity(vector1, vector2)
sim13 = cosine_similarity(vector1, vector3)
sim23 = cosine_similarity(vector2, vector3)

# df로 만드려고 딕셔너리로 집어넣음
result_df = pd.DataFrame({"Image1": ["img1","img1","img2"],
                          "Image2": ["img2","img3","img3"],
                          "Cosine Similarity":[sim12, sim13, sim23]})

# .round() : 소수점 정리
result_df["Cosine Similarity"]=(result_df["Cosine Similarity"].round(4))

result_df.sort_values(by="Cosine Similarity",ascending=False).reset_index(drop=True)

# 그냥 등수 이쁘게 맞추려고
result_df.index = result_df.index +1
result_df.index.name="Rank"

# 표 출력
display(result_df)

# 가장 유사한 이미지 : result_df.iloc[0]
```
